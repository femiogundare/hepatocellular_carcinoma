```r
# Which samples are primary solid tumor
dataSmTP <- TCGAquery_SampleTypes(getResults(query_TCGA,cols="cases"),"TP")
```

```r
# which samples are solid tissue normal
dataSmNT <- TCGAquery_SampleTypes(getResults(query_TCGA,cols="cases"),"NT")
```

```r
dataPrep <- TCGAanalyze_Preprocessing(tcga_data, cor.cut = 0.6, datatype=names(assays(tcga_data))[1])
```

```r
dataNorm <- TCGAanalyze_Normalization(tabDF = dataPrep, geneInfo =  geneInfoHT, method="gcContent")
```

```r
dataFilt <- TCGAanalyze_Filtering(tabDF = dataNorm,
                                  method = "quantile",
                                  qnt.cut =  0.25)
```

```r
dataDEGs <- TCGAanalyze_DEA(mat1 = dataFilt[,dataSmNT],
                            mat2 = dataFilt[,dataSmTP],
                            Cond1type = "Normal",
                            Cond2type = "Tumor",
                            fdr.cut = 0.01 ,
                            logFC.cut = 1.5,
                            method = "glmLRT")
```

```r
dataDEGs
```

```r
# Write values to text file using write.table
write.table(dataDEGs, file='degs.txt')
```

```r
dataDEGs$diffexpressed <- ifelse(dataDEGs$logFC > 0 & dataDEGs$PValue < 0.05, "UP",
                       ifelse(dataDEGs$logFC < 0 & dataDEGs$PValue < 0.05, "DOWN", "N0"))

dataDEGs$delabel <- "NA"
```

```r
table(dataDEGs$diffexpressed)
```

```r
accessionID <- "TCGA-LIHC"
```

```r
ggplot(data=dataDEGs, aes(x=logFC, y=-log10(PValue), col=diffexpressed, label=delabel)) +
    geom_point() +
    theme_minimal() +
    scale_color_manual(values=c("red", "green", "black")) +
    scale_x_continuous(breaks = seq(0, 6, by = 2), labels = c("0.0", "2.0", "4.0", "6.0")) +
    geom_vline(xintercept = c(-1.0,  1.0), col="red") +
    geom_hline(yintercept = 0.5, col="red") +
    theme(text = element_text(size=20))
```

```r
# Volcano plot
jpeg(file=sprintf("Volcano Plot - %s.jpeg", accessionID), width = 800, height = 800)
ggplot(data=dataDEGs, aes(x=logFC, y=-log10(PValue), col=diffexpressed, label=delabel)) +
    geom_point() +
    theme_minimal() +
    scale_color_manual(values=c("red", "green", "black")) +
    scale_x_continuous(breaks = seq(0, 6, by = 2), labels = c("0.0", "2.0", "4.0", "6.0")) +
    geom_vline(xintercept = c(-1.0,  1.0), col="red") +
    geom_hline(yintercept = 0.5, col="red") +
    theme(text = element_text(size=20))
dev.off()
```
