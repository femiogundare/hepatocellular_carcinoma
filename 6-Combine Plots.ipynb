```python
pdf_files = [
    "COL24A1_survival_curve.pdf",
    "RAD51_survival_curve.pdf",
    "CLSPN_survival_curve.pdf",
    "ECT2_survival_curve.pdf",
    "IQGAP3_survival_curve.pdf",
    "ARHGAP11A_survival_curve.pdf",
    "CCL23_survival_curve.pdf",
    "SNHG4_survival_curve.pdf",
    "PLK4_survival_curve.pdf",
    "DNAJC6_survival_curve.pdf",
    "BAIAP2L2_survival_curve.pdf",
    "PIF1_survival_curve.pdf",
    "FANCD2_survival_curve.pdf",
    "BUB1_survival_curve.pdf",
    "TMC7_survival_curve.pdf",
    "CENPI_survival_curve.pdf",
    "COL9A1_survival_curve.pdf",
    "COL25A1_survival_curve.pdf",
    "STRA6_survival_curve.pdf",
    "KIAA1841_survival_curve.pdf",
    "RANBP3L_survival_curve.pdf",
    "DIRAS2_survival_curve.pdf",
    "CREG2_survival_curve.pdf",
    "FANCB_survival_curve.pdf"
]
```

```python
import fitz
from  PIL import Image
import matplotlib.pyplot as plt
import os


def pdf_to_image(pdf_file):
    doc = fitz.open(pdf_file)
    page = doc.load_page(0)
    pix = page.get_pixmap()
    img = Image.frombytes("RGB", [pix.width, pix.height], pix.samples)
    return img

images = [pdf_to_image(pdf_file) for pdf_file in pdf_files]

rows = 5
cols = 5

fig, axes = plt.subplots(rows, cols, figsize=(15, 15))
axes = axes.flatten()

for i, (img, ax) in enumerate(zip(images, axes)):
    ax.imshow(img)
    ax.axis('off')

    ax.set_title(f'({chr(97 + i)})', fontsize=12, fontweight='bold')

for j in range(len(images), len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout(pad=3.0)
plt.savefig('combined_survival_curves_with_labels.png', dpi=300)
plt.show()
```

```python
import matplotlib.pyplot as plt
from PIL import Image

img_paths = [
    'confmatrix_train.png',
    'confmatrix_test.png',
    'roc_auc.png',
]

fig = plt.figure(figsize=(10, 10))
gs = fig.add_gridspec(2, 2, height_ratios=[1, 0.6], width_ratios=[1, 1], hspace=-0.1)

ax1 = fig.add_subplot(gs[0, 0])
ax2 = fig.add_subplot(gs[0, 1])
ax3 = fig.add_subplot(gs[1, :])

axes = [ax1, ax2, ax3]
for i, (ax, img_path) in enumerate(zip(axes, img_paths)):
    img = Image.open(img_path)
    ax.imshow(img)
    ax.axis('off')

    ax.text(
        0.5, 1.05, f"({chr(97 + i)})", color='black', fontsize=12,
        fontweight='bold', ha='center', va='bottom', transform=ax.transAxes
    )

plt.subplots_adjust(left=0.05, right=0.95, top=0.95, bottom=0.05)
plt.savefig('combined_metric_plot.png', dpi=300)
plt.show()
```
